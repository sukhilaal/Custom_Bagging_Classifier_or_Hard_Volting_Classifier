```python
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
X, y = make_classification(n_samples = 1000, n_features = 20, n_classes = 2, random_state = 1 )
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2, random_state=1)
```

```python
X_train.shape, X_test.shape
```

```python
from sklearn.ensemble import RandomForestClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.svm import SVC
from sklearn.ensemble import VotingClassifier
```

```python
from sklearn.linear_model import LogisticRegression
from sklearn.naive_bayes import GaussianNB

Llr_clf = LogisticRegression()
dt_clf = DecisionTreeClassifier()
sv_clf = SVC(
    probability=True, kernel='linear'
)
GNB_clf = GaussianNB()
```

```python
# for ensamblr
from sklearn.ensemble import VotingClassifier

voting_clf = VotingClassifier(
    estimators=[('lr', Llr_clf), ('dt', dt_clf), ('svc', sv_clf)],
    voting='soft'  # 'soft' for estimated probabilities
)
voting_clf.fit(X_train, y_train)

# Evaluate the voting classifier
from sklearn.metrics import accuracy_score
y_pred_voting = voting_clf.predict(X_test)
print(f"Voting Classifier Accuracy: {accuracy_score(y_test, y_pred_voting)}")
```

```python
# Train and evaluate individual models

print("--- Individual Model Performance ---")

# Logistic Regression
Llr_clf.fit(X_train, y_train)
y_pred_lr = Llr_clf.predict(X_test)
acc_lr = accuracy_score(y_test, y_pred_lr)
print(f"Logistic Regression Accuracy: {acc_lr:.4f}")

# Decision Tree
dt_clf.fit(X_train, y_train)
y_pred_dt = dt_clf.predict(X_test)
acc_dt = accuracy_score(y_test, y_pred_dt)
print(f"Decision Tree Accuracy: {acc_dt:.4f}")

# Support Vector Classifier (SVC)
sv_clf.fit(X_train, y_train)
y_pred_svc = sv_clf.predict(X_test)
acc_svc = accuracy_score(y_test, y_pred_svc)
print(f"SVC Accuracy: {acc_svc:.4f}")

print("\n--- Ensemble Model Performance ---")
# Voting Classifier (already fitted in previous cell, just re-evaluate for consistency)
y_pred_voting = voting_clf.predict(X_test)
acc_voting = accuracy_score(y_test, y_pred_voting)
print(f"Voting Classifier Accuracy: {acc_voting:.4f}")

print("\n--- Performance Summary (Tree-like conceptual structure) ---")
print("Root: Ensemble Model")
print(f"  └── Voting Classifier: {acc_voting:.4f}")
print("      ├── Base Model 1: Logistic Regression: ", f"{acc_lr:.4f}")
print("      ├── Base Model 2: Decision Tree:         ", f"{acc_dt:.4f}")
print("      └── Base Model 3: SVC:                   ", f"{acc_svc:.4f}")
```
